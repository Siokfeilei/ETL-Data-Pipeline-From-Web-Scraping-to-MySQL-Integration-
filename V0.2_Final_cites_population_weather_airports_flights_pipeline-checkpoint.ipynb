```python
#PART 4 — Fetch Weather Data (NON-STATIC) -->This function retrieves forecasts for tomorrow from OpenWeatherMap, using lat/lon stored in cities.


import pandas as pd
import requests
from pytz import timezone
from datetime import datetime
from sqlalchemy import create_engine

# --- Create MySQL connection ---
def create_connection_string():
    schema = "gans_local_database"
    host = "127.0.0.1"
    user = "root"
    password = "GK_th3worst!"
    port = 3307
    return f'mysql+pymysql://{user}:{password}@{host}:{port}/{schema}'

# --- Fetch cities from database ---
def fetch_cities_data(connection_string):
    engine = create_engine(connection_string)
    return pd.read_sql("SELECT * FROM cities", con=engine)

# --- Fetch weather data ---
def fetch_weather_data(cities_df):
    berlin_timezone = timezone('Europe/Berlin')
    API_key = "6cd3e3648fc414e4db1760982685383f"
    weather_items = []

    for _, city in cities_df.iterrows():
        latitude = city["latitude"]
        longitude = city["longitude"]
        city_id = city["city_id"]

        url = (
            f"https://api.openweathermap.org/data/2.5/forecast"
            f"?lat={latitude}&lon={longitude}&appid={API_key}&units=metric"
        )

        response = requests.get(url)
        data = response.json()

        retrieval_time = datetime.now(berlin_timezone)

        for item in data.get("list", []):
            weather_items.append({
                "city_id": city_id,
                "forecast_datetime": item.get("dt_txt"),
                "collected_datetime": retrieval_time,
                "temp_Celsius": item["main"].get("temp"),
                "feels_like_Celsius": item["main"].get("feels_like"),
                "humidity": item["main"].get("humidity"),
                "wind_speed_mps": item["wind"].get("speed"),
                "weather_desc": item["weather"][0].get("main"),
                "precipitation_probability": item.get("pop", 0),
                "rain_in_last_3h": item.get("rain", {}).get("3h", 0)
            })

    weather_df = pd.DataFrame(weather_items)
    weather_df["forecast_datetime"] = pd.to_datetime(weather_df["forecast_datetime"])
    weather_df["collected_datetime"] = pd.to_datetime(weather_df["collected_datetime"])
    return weather_df
```

```python
# --- Store weather data ---
def store_weather_data(weather_df, connection_string):
    engine = create_engine(connection_string)
    weather_df.to_sql('weather', con=engine, if_exists='append', index=False)
    print("Weather data appended successfully.")
```

```python
# --- Main function ---
def retrieve_and_send_data():
    connection_string = create_connection_string()
    cities_df = fetch_cities_data(connection_string)
    weather_df = fetch_weather_data(cities_df)
    store_weather_data(weather_df, connection_string)
    return "Weather data has been updated successfully."

```

```python
#PART 5 — Fetch Airports (STATIC) using cities table 
import pandas as pd
import requests
from sqlalchemy import create_engine

# --- MySQL connection ---
def create_connection():
    user = "root"
    password = "GK_th3worst!"  # <- replace with your MySQL password
    host = "127.0.0.1"
    port = 3307
    db = "gans_local_database"
    return create_engine(f"mysql+pymysql://{user}:{password}@{host}:{port}/{db}")

# --- Fetch airports for each city ---
def fetch_airports(api_key):
    engine = create_connection()
    
    # Read cities from MySQL
    cities = pd.read_sql("SELECT * FROM cities", engine)
    
    # Fetch existing ICAO codes to avoid duplicates
    existing_icao_codes = pd.read_sql("SELECT icao_code FROM airports", engine)['icao_code'].tolist()
    
    airport_rows = []

    for _, city in cities.iterrows():
        lat = city['latitude']
        lon = city['longitude']
        city_id = city['city_id']

        url = f"https://aerodatabox.p.rapidapi.com/airports/search/location/{lat}/{lon}/km/50/16"
        headers = {
            "X-RapidAPI-Key": api_key,
            "X-RapidAPI-Host": "aerodatabox.p.rapidapi.com"
        }

        response = requests.get(url, headers=headers)
        if response.status_code != 200:
            print(f"Error fetching airports for {city['city']}: {response.status_code}")
            continue

        data = response.json()
        for airport in data.get("items", []):
            icao_code = airport.get("icao")
            if icao_code and icao_code not in existing_icao_codes:
                airport_rows.append({
                    "icao_code": icao_code,
                    "name": airport.get("name"),
                    "city_id": city_id,
                    "latitude": airport.get("location", {}).get("lat"),
                    "longitude": airport.get("location", {}).get("lon")
                })
                # Add to existing codes list to prevent duplicates in same run
                existing_icao_codes.append(icao_code)

    # Create DataFrame
    df_airports = pd.DataFrame(airport_rows)

    # Append to MySQL only if there are new airports
    if not df_airports.empty:
        df_airports.to_sql("airports", engine, if_exists="append", index=False)
        print(f"Inserted {len(df_airports)} new airports.")
    else:
        print("No new airports to insert.")
```

```python
#PART 6 — Fetch Flights for Tomorrow (NON-STATIC) -->fetchse the stored airport ICAO codes to retrieve tomorrow’s arrivals.
import pandas as pd
import requests
from datetime import datetime, timedelta
import pytz
from sqlalchemy import create_engine

BERLIN_TZ = pytz.timezone("Europe/Berlin")

def create_connection():
    user = "root"
    password = "GK_th3worst!"  
    host = "127.0.0.1"
    port = 3307
    db = "gans_local_database"
    return create_engine(f"mysql+pymysql://{user}:{password}@{host}:{port}/{db}")

def fetch_flights_to_mysql():
    engine = create_connection()
    tomorrow = (datetime.now(BERLIN_TZ) + timedelta(days=1)).date()
    airports = pd.read_sql("SELECT icao_code FROM airports", engine)
    flight_rows = []

    print(f"Fetching flights for {len(airports)} airports on {tomorrow}")

    # Split day into two 12-hour periods
    times = [["00:00","11:59"], ["12:00","23:59"]]

    for icao in airports['icao_code']:
        for time in times:
            start_dt = datetime.combine(tomorrow, datetime.strptime(time[0], "%H:%M").time()).strftime("%Y-%m-%dT%H:%M:%S")
            end_dt = datetime.combine(tomorrow, datetime.strptime(time[1], "%H:%M").time()).strftime("%Y-%m-%dT%H:%M:%S")

            url = f"https://aerodatabox.p.rapidapi.com/flights/airports/icao/{icao}/{start_dt}/{end_dt}"
            headers = {
                'x-rapidapi-host': "aerodatabox.p.rapidapi.com",
                'x-rapidapi-key': "687292277emsh6620811a3972b04p1a4ee9jsn8c02f9bc139b"
            }
            params = {
                "withLeg": "true",
                "direction": "Arrival",
                "withCancelled": "false",
                "withCodeshared": "true",
                "withCargo": "false",
                "withPrivate": "false"
            }

            try:
                response = requests.get(url, headers=headers, params=params)
                response.raise_for_status()
                data = response.json()
            except Exception as e:
                print(f"Error fetching flights for {icao}: {e}")
                continue

            arrivals = data.get("arrivals", [])
            if not arrivals:
                continue

            for f in arrivals:
                flight_rows.append({
                    "arrival_airport_icao": icao,
                    "departure_airport_icao": f["departure"]["airport"].get("icao"),
                    "scheduled_arrival_time": f["arrival"]["scheduledTime"].get("local"),
                    "flight_number": f.get("number"),
                    "data_retrieved_at": datetime.now(BERLIN_TZ)
                })

    if flight_rows:
        df_flights = pd.DataFrame(flight_rows)

        # Remove duplicates before inserting
        df_flights = df_flights.drop_duplicates(
            subset=["arrival_airport_icao", "flight_number", "scheduled_arrival_time"]
        )

        # Convert to proper datetime
        df_flights["scheduled_arrival_time"] = pd.to_datetime(df_flights["scheduled_arrival_time"])
        df_flights["data_retrieved_at"] = pd.to_datetime(df_flights["data_retrieved_at"])

        print(f"Saving {len(df_flights)} unique flights to MySQL...\n")
        
        # Print simple step-by-step flight codes
        for idx, row in df_flights.iterrows():
            print(f"{idx+1}. {row['arrival_airport_icao']} - {row['flight_number']} - {row['scheduled_arrival_time']}")

        try:
            df_flights.to_sql("flights", engine, if_exists="append", index=False)
            print("\nFlights successfully saved to MySQL!")
        except Exception as e:
            print(f"Error saving flights to MySQL: {e}")
    else:
        print("No flights to save.")

```
